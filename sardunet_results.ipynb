This notebook will load some results from **SARDU-Net** and tell you the best measurements to use

```python
import numpy as np

# Results are here
dir_results = "sarduout/train_nnsel126-71-126_psel0.2_dsel16_nnpred126-71-126_ppred0.2_noepoch100_lr0.001_mbatch32_seed257891_lossL2_prct99.0_small1e-06/"

# Load a vector the best measurements
list_best = np.loadtxt(dir_results + "nnet_lossvalmin_measidx.txt", dtype=int)

# Load the parameter lists
param_lists = np.load("input_data/sampled_parameters_M-126.npz")
```

Now go through the best measurement indices and print the combinations of $b$, $\delta$, and $\Delta$ that the network has found

```python
print("The best measurements correspond to the following parameter combinations:")
for idx in list_best:
    b_val = param_lists['b'][idx]
    delta_val = param_lists['delta'][idx]
    sep_val = param_lists['sep'][idx]
    print(f"Index: {idx}, b: {b_val}, delta: {delta_val}, sep: {sep_val}")
```
